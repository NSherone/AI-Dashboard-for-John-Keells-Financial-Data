```python
import os
import camelot
import pandas as pd

def extract_data_from_pdf(pdf_path, pages_to_extract, output_folder, start_index=1):
   
    os.makedirs(output_folder, exist_ok=True)

    extracted_data = {}

    for label, page in pages_to_extract.items():
        print(f"Extracting '{label}' from page {page} of '{pdf_path}'")
        tables = camelot.read_pdf(pdf_path, pages=str(page), flavor='stream')

        if tables:
            extracted_data[label] = [t.df for t in tables]
        else:
            print(f"No tables found on page {page}.")

    for label, dfs in extracted_data.items():
        for idx, df in enumerate(dfs, start=start_index):
            filename = f"{label}_table_{idx}.csv"
            csv_path = os.path.join(output_folder, filename)
            df.to_csv(csv_path, index=False)
            print(f"Saved: {csv_path}")

# Configuration for multiple PDFs
pdf_files = [
    {
        "pdf_path": "2020.pdf",
        "pages_to_extract": {
            "top_20_shareholders": 150,
            "income_statement": 184,
            "net_asset_ebitda": 275,
            "equity_liabilities": 186
        },
        "output_folder": "extracted_csvs/2020",
        "start_index": 2020
    },
    {
        "pdf_path": "2021.pdf",
        "pages_to_extract": {
            "top_20_shareholders": 162,
            "income_statement": 198,
            "net_asset_ebitda": 287,
            "equity_liabilities": 200
        },
        "output_folder": "extracted_csvs/2021",
        "start_index": 2021
    },
    
    {
        "pdf_path": "2022.pdf",
        "pages_to_extract": {
            "top_20_shareholders": 172,
            "income_statement": 208,
            "net_asset_ebitda": 297,
            "equity_liabilities": 210
        },
        "output_folder": "extracted_csvs/2022",
        "start_index": 2022
    },
    {
        "pdf_path": "2023.pdf",
        "pages_to_extract": {
            "top_20_shareholders": 172,
            "income_statement": 216,
            "net_asset_ebitda": 307,
            "equity_liabilities": 218
        },
        "output_folder": "extracted_csvs/2023",
        "start_index": 2023
    },
    {
        "pdf_path": "2024.pdf",
        "pages_to_extract": {
            "top_20_shareholders": 131,
            "income_statement": 276,
            "net_asset_ebitda": 369,
            "equity_liabilities": 278
        },
        "output_folder": "extracted_csvs/2024",
        "start_index": 2024
    }
]

for pdf_file in pdf_files:
    extract_data_from_pdf(
        pdf_path=pdf_file["pdf_path"],
        pages_to_extract=pdf_file["pages_to_extract"],
        output_folder=pdf_file["output_folder"],
        start_index=pdf_file.get("start_index", 1)
    )
```

```python
import pandas as pd
import re


year_config = {
    "2019": {"income_cols": [3], "other_cols": [2], "tables": {
        "income": income_statement_table_2020,
        "net_asset": net_asset_ebitda_table_2020,
        "equity": equity_liabilities_table_2020,
    }},
    "2020": {"income_cols": [2], "other_cols": [1], "tables": {
        "income": income_statement_table_2020,
        "net_asset": net_asset_ebitda_table_2020,
        "equity": equity_liabilities_table_2020,
    }},
    "2021": {"income_cols": [2], "other_cols": [1], "tables": {
        "income": income_statement_table_2021,
        "net_asset": net_asset_ebitda_table_2021,
        "equity": equity_liabilities_table_2021,
    }},
    "2022": {"income_cols": [2], "other_cols": [1], "tables": {
        "income": income_statement_table_2022,
        "net_asset": net_asset_ebitda_table_2022,
        "equity": equity_liabilities_table_2022,
    }},
    "2023": {"income_cols": [2], "other_cols": [1], "tables": {
        "income": income_statement_table_2023,
        "net_asset": net_asset_ebitda_table_2023,
        "equity": equity_liabilities_table_2023,
    }},
    "2024": {"income_cols": [2], "other_cols": [1], "tables": {
        "income": income_statement_table_2024,
        "net_asset": net_asset_ebitda_table_2024,
        "equity": equity_liabilities_table_2024,
    }},
}


metrics = [
    "Total Revenue", "Cost of Sales", "Gross Profit", "Operating Expenses",
    "Basic EPS", "Diluted EPS", "Net assets per share",  "Total equity and liabilities" ,"Net Profit"
]


extracted = {metric: {} for metric in metrics}


key_mapping = {
    "total revenue": "Total Revenue",
    "cost of sales": "Cost of Sales",
    "gross profit": "Gross Profit",
    "basic": "Basic EPS",
    "diluted": "Diluted EPS",
    "net assets per share": "Net assets per share",
    "total equity and liabilities": "Total equity and liabilities",
    "profit for the year": "Net Profit"
}


expense_keywords = [
    "selling and distribution expenses",
    "administrative expenses",
    "other operating expenses"
]

def clean_value(val):
    val = str(val).replace(",", "").replace("(", "-").replace(")", "")
    return float(val) if val.lower() not in ["nan", "-", ""] else None

def extract_from_df(df, col_indices, year, match_eps_only=False, multiplier=1):
    for _, row in df.iterrows():
        row_text = " ".join(map(str, row)).lower()
        try:
            values = [clean_value(row.iloc[i]) for i in col_indices]
        except:
            values = [None] * len(col_indices)

        matched = False
        for keyword, label in key_mapping.items():
           
            if re.search(rf"\b{re.escape(keyword)}\b", row_text):
                value = values[0]
                if value is not None:
                    extracted[label][year] = value 
                matched = True
                break

        
        if not matched and any(kw in row_text for kw in expense_keywords):
            value = values[0]
            if value is not None:
                extracted["Operating Expenses"][year] = extracted["Operating Expenses"].get(year, 0.0) + value

def extract_specific(df, col_indices, year):
    for _, row in df.iterrows():
        row_text = " ".join(map(str, row)).lower()
        try:
            values = [clean_value(row.iloc[i]) for i in col_indices]
        except:
            values = [None] * len(col_indices)

        if "total equity and liabilities" in row_text:
            value = values[0]
            if value is not None:
                extracted["Total equity and liabilities"][year] = value

for year, config in year_config.items():
    extract_from_df(config["tables"]["income"], config["income_cols"], year, match_eps_only=False)
    extract_from_df(config["tables"]["net_asset"], config["other_cols"], year, match_eps_only=True)
    extract_specific(config["tables"]["equity"], config["income_cols"], year)


result_df = pd.DataFrame(extracted).T
result_df.loc["Basic EPS", "2021"] =  3.62 
result_df.loc["Diluted EPS", "2021"] =  3.62 
result_df.loc["Basic EPS", "2022"] = 15.13
result_df.loc["Diluted EPS", "2022"] = 15.12
result_df.loc["Basic EPS", "2023"] =  13.12 
result_df.loc["Diluted EPS", "2023"] =  13.12 
formatted_df = result_df.copy()

for col in formatted_df.columns:
    formatted_df[col] = formatted_df[col].map(
        lambda x: f"{x:,.2f}" if isinstance(x, (int, float)) and pd.notnull(x) else x
    )

formatted_df
formatted_df.to_json("formatted_financial_data.json", orient="index", indent=4)
```

```python
def generate_ml_insight(df, selected_year):
    selected_year = str(selected_year)
    
    if selected_year not in df.columns:
        return [f" No financial data available for year {selected_year}"]


    years = list(df.columns)
    year_idx = years.index(selected_year)

    if year_idx == 0:
        return ["ℹ️ Not enough historical data for comparison."]
    
    prev_year = years[year_idx - 1]
    insights = []

    emoji_map = {
        "Total Revenue": "💼",
        "Cost of Sales": "📦",
        "Gross Profit": "📊",
        "Operating Expenses": "🧾",
        "Net Profit": "💰",
        "Basic EPS": "📉",
        "Diluted EPS": "📉",
        "Net assets per share": "🏦",
        "Total equity and liabilities": "🏛️"
    }

    for metric in df.index:
        try:
            current = df.at[metric, selected_year]
            previous = df.at[metric, prev_year]
            change = current - previous
            pct_change = (change / abs(previous)) * 100 if previous != 0 else 0

            tag = emoji_map.get(metric, "📈")

            if pct_change > 10:
                insights.append(f"{tag} {metric} increased by {pct_change:.2f}% in {selected_year} compared to {prev_year}.")
            elif pct_change < -10:
                insights.append(f"{tag} {metric} dropped by {abs(pct_change):.2f}% in {selected_year} compared to {prev_year}.")
            elif abs(pct_change) > 2:
                direction = "rose" if change > 0 else "fell"
                insights.append(f"{tag} {metric} slightly {direction} ({pct_change:.2f}%) in {selected_year} vs {prev_year}.")
        except Exception as e:
            insights.append(f"⚠️ Could not compute insight for {metric}: {str(e)}")

    return insights
```

```python
import fitz  
import os

def extract_text_from_pdf(pdf_path, output_txt_path):
    try:
        os.makedirs(os.path.dirname(output_txt_path), exist_ok=True)
        doc = fitz.open(pdf_path)
        full_text = "\n".join([page.get_text() for page in doc])
        with open(output_txt_path, "w", encoding="utf-8") as f:
            f.write(full_text)
        print(f" Saved: {output_txt_path}")
    except Exception as e:
        print(f" Failed for {pdf_path}: {e}")


years = ["2020", "2021", "2022", "2023", "2024"]


for year in years:
    pdf_filename = f"{year}.pdf" 
    output_filename = f"text_output/{year}_report.txt"
    extract_text_from_pdf(pdf_filename, output_filename)
```

```python
import os
import json
import requests

API_KEY = "API_KEY"

def call_openrouter_llm(summary_text):
    prompt = f"""
    From the following annual report excerpt, identify major events or economic disruptions that impacted the company's financial performance. 
    List each event as a bullet point. If no events are mentioned, return an empty list.

    Text: {summary_text}

    Event Annotations:
    """

    url = "https://openrouter.ai/api/v1/chat/completions"
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json",
    }

    data = json.dumps({
        "model": "deepseek/deepseek-r1:free",
        "messages": [
            {
                "role": "user",
                "content": prompt.strip()
            }
        ]
    })

    response = requests.post(url, headers=headers, data=data)
    if response.status_code == 200:
        content = response.json()['choices'][0]['message']['content']
        return content.strip().split('\n')
    else:
        print(f"Error {response.status_code}: {response.text}")
        return []

text_folder = "text_output"
years = ["2020", "2021", "2022", "2023", "2024"]
event_annotations = {}

for year in years:
    try:
        with open(f"{text_folder}/{year}_report.txt", "r", encoding="utf-8") as f:
            report_text = f.read()

        print(f"Extracting events for {year}...")
        year_events = call_openrouter_llm(report_text[:3000])  
        event_annotations[year] = [e.strip("•- ") for e in year_events if e.strip()]
    except Exception as e:
        print(f"Failed to process {year}: {e}")
        event_annotations[year] = []


with open("event_annotations.json", "w", encoding="utf-8") as f:
    json.dump(event_annotations, f, indent=2)

print("Saved event_annotations.json")
```
